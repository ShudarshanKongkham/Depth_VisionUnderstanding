# 1. Initialize RealSense Pipeline and Stream Configuration

In this section, we'll import the required library, create a RealSense pipeline, and configure it to stream both depth and color frames at 640x480 resolution and 30 FPS.

```python
import pyrealsense2 as rs

# Create a pipeline
pipeline = rs.pipeline()

# Create a config and configure the pipeline to stream
config = rs.config()
config.enable_stream(rs.stream.depth, 640, 480, rs.format.z16, 30)
config.enable_stream(rs.stream.color, 640, 480, rs.format.bgr8, 30)
```

# 2. Acquire Depth and Color Frames

Start the pipeline and acquire synchronized depth and color frames using `wait_for_frames()`.

```python
# Start streaming
profile = pipeline.start(config)

# Wait for a coherent pair of frames: depth and color
frames = pipeline.wait_for_frames()
depth_frame = frames.get_depth_frame()
color_frame = frames.get_color_frame()

if not depth_frame or not color_frame:
    raise RuntimeError("Could not acquire depth or color frames.")
```

# 3. Visualize Depth Frame

Convert the depth frame to a NumPy array and display it using matplotlib's `imshow` with a colormap (e.g., 'plasma').

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert depth frame to numpy array
depth_image = np.asanyarray(depth_frame.get_data())

plt.figure(figsize=(8, 6))
plt.imshow(depth_image, cmap='plasma')
plt.title('Depth Frame')
plt.colorbar(label='Depth Value')
plt.axis('off')
plt.show()
```

# 4. Visualize Color Frame

Convert the color frame to a NumPy array and display it using matplotlib's `imshow`.

```python
# Convert color frame to numpy array
color_image = np.asanyarray(color_frame.get_data())

plt.figure(figsize=(8, 6))
plt.imshow(color_image)
plt.title('Color Frame')
plt.axis('off')
plt.show()
```

# 5. Display Depth Value at Image Center

Calculate the center pixel coordinates and print the depth value at that location in meters.

```python
height, width = depth_image.shape
center_x = int(width / 2)
center_y = int(height / 2)
center_depth = depth_frame.get_distance(center_x, center_y)
print(f"Depth at center ({center_x}, {center_y}): {center_depth:.2f} meters")
```

# 6. List Available Cameras

Query connected RealSense devices and print their names and information.

```python
context = rs.context()
devices = context.query_devices()
print("Available RealSense cameras:")
for i, device in enumerate(devices):
    name = device.get_info(rs.camera_info.name)
    serial = device.get_info(rs.camera_info.serial_number)
    print(f"Camera {i+1}: {name} (Serial: {serial})")
```

```python
# Stop streaming when done
pipeline.stop()
```
